# Loading Alignment Files

```r
dir <- "/mmfs1/gscratch/stergachislab/mhsohny/SMaHT/Improving_SomaticVariantCalling_through_DSA/Revision/CDR_reads_to_other_asm"
colo829bl_dsa_cdr_bam <- paste(dir, "reads_overlapping_cdrs_from_COLO829BL_DSA_resetmapq.bam", sep="/") 
colo829bl_chm13_cdr_bam <- paste(dir, "chm13/reads_overlapping_cdrs_from_COLO829BL_DSA_resetmapq_chm13.bam", sep="/")
colo829bl_hg38_cdr_bam <- paste(dir, "hg38/reads_overlapping_cdrs_from_COLO829BL_DSA_resetmapq_hg38.bam", sep="/")
```

# Karyoplot of COLO829BL DSA

```r
options(repr.plot.width = 40, repr.plot.height = 5)
par(bg="white")

contig <- "haplotype1-0000012"
region <- GRanges(seqnames = contig, ranges = IRanges(start = 122887530, end = 123653747))

n.bins     <- 4000
binsize    <- ceiling(width(region) / n.bins)
aln        <- readGAlignments(colo829bl_dsa_cdr_bam, param = ScanBamParam(which = region))
cov        <- coverage(aln)[[contig]]
bin.starts <- seq(start(region), end(region) - binsize + 1L, by = binsize)
bin.cov    <- viewMeans(Views(cov, start = bin.starts, width = binsize))
ymax       <- max(bin.cov)

pdf(paste(dir, "colo829bl_chr1_hap1_cdr_coverage.pdf", sep = "/"), width = 40, height = 5, bg = "white")

kp <- plotKaryotype(genome = dsa, plot.type=4, cex=3, chromosomes=contig, zoom=region)
kpAddBaseNumbers(kp, tick.dist = 100000, cex=2, add.units = TRUE)
kp <- kpArea(kp, chr=contig, x=bin.starts + binsize/2, y=bin.cov,
             base.y=0, ymin=0, ymax=ymax, col="gray60", border=NA)
kpAxis(kp, ymax=ymax, cex=2)

dev.off()
```
